## Пять понятий

**Масштабирование** меняет количество пикселей. Коэффициент `0.5` уменьшает обе стороны вдвое: изображение 800×600 становится 400×300. При уменьшении программа объединяет информацию соседних пикселей. Обратное увеличение не возвращает потерянные мелкие детали.

**Вырезание, или crop**, выбирает прямоугольную область. Четыре числа `X Y W H` задают её левый верхний угол, ширину и высоту. Выбранная область получает собственное начало координат: её прежняя точка `(X, Y)` становится `(0, 0)`.

**Сдвиг и отражение** меняют положение содержимого. Положительный `DX` передвигает изображение вправо, положительный `DY` — вниз. Горизонтальное отражение переставляет левую и правую стороны; это важно помнить при интерпретации картинки с камеры.

**Аффинное преобразование** включает линейное изменение координат и перенос. В примере добавлен небольшой наклон, или сдвиг строк относительно друг друга. Прямые сохраняются прямыми, параллельные направления остаются параллельными.

**Перспективное преобразование** сопоставляет четыре угла плоского объекта с углами нового прямоугольника. Оно помогает представить наклонённую табличку как вид спереди. Это работа с одной плоскостью; объекты вне неё могут выглядеть искажёнными. Само преобразование не определяет расстояние до объекта в метрах.

8. Измените размер результата плоскости и сравните пропорции:

   ```bash
   python3 -m rosmaster_cv.lesson02_geometry --input data/portrait_plane.png --warp-size 400 400 --out outputs/02_square
   ```

   Объясните, почему квадратный результат растягивает содержимое по сравнению с исходными пропорциями таблицы.

## Ячейки практики

Запускайте последовательно. Команды используют тот же Python, что и ядро ноутбука. Рабочая папка дочерних процессов — корень репозитория. Ошибка команды останавливает ячейку и показывает причину.


````python
from pathlib import Path
import subprocess
import sys

# Jupyter may start in the repository root or in its notebooks directory.
ROOT = None
for candidate in (Path.cwd().resolve(), Path.cwd().resolve().parent):
    if (candidate / "rosmaster_cv").is_dir() and (candidate / "lessons").is_dir():
        ROOT = candidate
        break
if ROOT is None:
    raise RuntimeError("Откройте ноутбук из корня репозитория или его папки notebooks.")

def run_module(module, *arguments):
    command = [sys.executable, "-m", "rosmaster_cv." + module]
    command.extend(str(argument) for argument in arguments)
    return subprocess.run(command, cwd=str(ROOT), check=True)

def run_tool(filename, *arguments):
    command = [sys.executable, str(ROOT / "tools" / filename)]
    command.extend(str(argument) for argument in arguments)
    return subprocess.run(command, cwd=str(ROOT), check=True)

print("Корень курса:", ROOT)
print("Python ядра:", sys.executable)
````

````python
run_tool("generate_data.py")
run_module("lesson02_geometry", "--input", "data/scene.png", "--out", "outputs/02")
````

````python
def show_results(relative_directory, limit=4):
    directory = ROOT / relative_directory
    paths = sorted(directory.glob("*.png"))[:limit]
    if not paths:
        print("Изображений пока нет:", directory)
        return
    try:
        from IPython.display import Image, display
    except ImportError:
        for path in paths:
            print(path)
        return
    for path in paths:
        print(path.relative_to(ROOT))
        display(Image(filename=str(path)))

show_results("outputs/02")
````
